# SpecConfig wizard (prototype)

An interactive `ipywidgets` form for assembling and reviewing a `SpecConfig` — the
single authoritative input to processing (see `docs/spec-config-inventory.md`).

**How to use it:**
1. Pick a **Model** (from `catalog/ModelSpec/`).
2. Optionally pick a **Domain** (from `catalog/DomainSpec/`) to prefill the grid,
   boundaries, partitioning, and dates — or leave it `<custom>` and enter your own.
3. Tweak any field. The **Review** pane re-resolves live via
   `build_spec_config(...)` and shows the resulting YAML (or a validation error).
4. *(optional)* click **Compute dt (CFL)** to fill `dt` from the grid (needs
   `roms_tools`); otherwise the `dt` field is used as-is.
5. **Save** writes `spec_config.yml`. Take that file to your machine of choice to
   run the processing step.

The wizard is a thin shell over `build_spec_config` — all resolution/validation
lives in `cstar_forge.spec_config_resolve`.

```python
from cstar_forge.spec_config_wizard import SpecConfigWizard

wiz = SpecConfigWizard()
wiz.display()
```

The resolved config is available as `wiz.config` (a `SpecConfig`, or `None` while
inputs are invalid). For example:

```python
cfg = wiz.config
if cfg is not None:
    print("name:    ", cfg.name)
    print("casename:", cfg.casename)
    print("dt:      ", cfg.model_settings["time_stepping"]["dt"])
    print("sources: ", list(cfg.sources.resolved_datasets))
```
