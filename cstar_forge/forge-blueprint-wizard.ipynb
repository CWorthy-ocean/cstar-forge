# ForgeBlueprint wizard (prototype)

An interactive `ipywidgets` form for assembling and reviewing a `ForgeBlueprint` — the
single authoritative input to processing (see `docs/dev-notes/forge-blueprint-inventory.md`).

**How to use it:**
1. *(optional)* Enter a different **Catalog** location (a local path, `"local"`, a
   GitHub URL, or an http URL) and click **Reload catalog** — the bundled in-repo
   catalog (`catalog/`) loads automatically otherwise.
2. Pick a **Model** (from `catalog/ModelSpec/`).
3. Optionally pick a **Domain** (from `catalog/DomainSpec/`) to prefill the grid,
   boundaries, partitioning, and dates — or leave it `<custom>` and enter your own.
4. Tweak any field. The **Review** pane re-resolves live via
   `build_forge_blueprint(...)` and shows the resulting YAML (or a validation error).
5. *(optional)* click **Compute dt (CFL)** to fill `dt` from the grid (needs
   `roms_tools`); otherwise the `dt` field is used as-is.
6. **Save** writes `forge_blueprint.yaml`. Take that file to your machine of choice to
   run the processing step.

The wizard is a thin shell over `build_forge_blueprint` — all resolution/validation
lives in `cstar_forge.forge_blueprint_resolve`.
